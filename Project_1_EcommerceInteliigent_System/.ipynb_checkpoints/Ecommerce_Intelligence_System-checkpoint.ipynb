```python

orders_data.isna().sum()#A lot of missing values
products_data.isna().sum()# alot missing values

dataframes = {
    "category_data": category_data,
    "customer_data": customer_data,
    "location_data": location_data,
    "order_item_data": order_item_data,
    "orders_data": orders_data,
    "payments_data": payments_data,
    "products_data": products_data,
    "reviews_data": reviews_data
}

for name, df in dataframes.items():
    print(f"\n{name} missing count:")
    print(df.isna().sum())
```

```python
#handle missing values properly
#product_weight_g                2
#product_length_cm               2
#product_height_cm               2
#product_width_cm                2
#these 4 seem very small can simply drop them

products_data = products_data.dropna(subset=['product_weight_g', 'product_length_cm', 'product_width_cm', 'product_height_cm'])

# product_category_name fill with unknown
products_data['product_category_name'] = products_data['product_category_name'].fillna('unknown')

# product_photos_qty is number, we can fill with 0
products_data['product_photos_qty'] = products_data['product_photos_qty'].fillna(0)

# product_name_lenght and product_description_lenght for empty results, fill with 0, 

products_data['product_name_lenght'] = products_data['product_name_lenght'].fillna(0)
products_data['product_description_lenght'] = products_data['product_description_lenght'].fillna(0)

display(products_data.head())
display(products_data.info())
display(products_data.isna().sum())
```

```python
#location_data_cleaned.shape
#location_data.shape

#Convert date columns to datetime format 

order_item_data['shipping_limit_date']=pd.to_datetime(order_item_data['shipping_limit_date'])
order_item_data.head()

reviews_data['review_creation_date']=pd.to_datetime(reviews_data['review_creation_date'])
reviews_data['review_answer_timestamp']=pd.to_datetime(reviews_data['review_answer_timestamp'])
reviews_data.dtypes

reviews_data.head(100)
orders_data.head()
date_columns=['order_purchase_timestamp','order_approved_at','order_delivered_carrier_date','order_delivered_customer_date','order_estimated_delivery_date']

for col in date_columns:
    orders_data[col]=pd.to_datetime(orders_data[col])


#orders data types
display(orders_data.dtypes)
display(orders_data.info())
display(orders_data.head())


#reviews data types
display(reviews_data.dtypes)
display(reviews_data.info())
display(reviews_data.head())

    

```

```python
#Standardize column names if required 

for name,dataframe in dataframes.items():

    dataframe.columns = (dataframe.columns
                  .str.strip()
                  .str.lower()
                  .str.replace(" ","_")
                  .str.replace("lenght","length"))

    print(f"Column names standardized for dataset: {name} => {dataframe.columns}")
```

```python
payments_agg = (
    payments_data
    .groupby('order_id')['payment_value']
    .sum()
    .reset_index()
    .rename(columns={'payment_value': 'total_order_value'})
)
```

```python
##Merge the data
#orders + customers 
master_table_df=orders_data.merge(customer_data, on='customer_id', how='left')

#orders + order_items 

master_table_df=master_table_df.merge(order_item_data,on ='order_id',how='left')

#order_items + products 
master_table_df=master_table_df.merge(products_data, on='product_id',how='left')

#orders + payments
master_table_df = master_table_df.merge(payments_agg,on='order_id',how='left')


#master_table_df=master_table_df.merge(payments_data, on='order_id',how='left')

#orders + reviews 
master_table_df=master_table_df.merge(reviews_data, on='order_id',how='left')


#order_items + sellers 

master_table_df=master_table_df.merge(sellers_data, on='seller_id',how='left')
#products + category_translation

master_table_df=master_table_df.merge(category_data, on='product_category_name',how='left')

display(master_table_df.head(10))
#display(master_table_df.info())
#display(master_table_df.shape)
#display(master_table_df['order_id'].duplicated().sum())
#display(order_customer_data)
#display(customer_data)
```

```python
print(master_table_df['order_id'].nunique())
print(len(master_table_df))
master_table_df['total_value'] = (master_table_df['price'] + master_table_df['freight_value'])
#print(master_table_df[['price', 'freight_value', 'total_order_value']].head())
print(master_table_df.info())
display(master_table_df.head())
print('total_value' in master_table_df.columns)
```

```python
#order_delivered_customer_date -order_purchase_timestamp should give the delivery time for a product

display(master_table_df['order_delivered_customer_date'].isna().sum())
display(master_table_df['order_purchase_timestamp'].isna().sum())

master_table_df['total_delivery_time'] = (master_table_df['order_delivered_customer_date']-master_table_df['order_purchase_timestamp'])
master_table_df['total_delivery_time_days'] = master_table_df['total_delivery_time'].dt.days

#master_table_df
display(master_table_df.head(10))
display(master_table_df.info())
display(master_table_df['order_delivered_customer_date'].dtype) 
display(master_table_df['order_purchase_timestamp'].dtype) 

```

```python
#Number of items per order 
master_table_df['items_per_order']=master_table_df.groupby('order_id')['order_item_id'].transform('count')
display(master_table_df)
master_table_df['items_per_order'].isna().sum()                                             
```

```python
#Customer purchase frequency 
#With customer_purchase_freq we can easily identify how many repeat customers been there.
#One-time customers (purchase_freq = 1)
#Repeat customers (purchase_freq > 1)
#Loyal customers (high purchase frequency)

#master_table_df['customer_purchase_freq']=master_table_df.groupby('customer_id')['order_id'].transform('count')

master_table_df['customer_purchase_freq'] = (
    master_table_df.groupby('customer_unique_id')['order_id']
    .transform('nunique')
)
display(master_table_df.head(10))
```

```python
#Customer lifetime value (basic approximation) -How much total money a customer brings to the business over time
master_table_df['customer_lifetime_value'] = master_table_df.groupby('customer_unique_id')['total_value'].transform('sum')
display(master_table_df.head(20))
display(master_table_df['customer_unique_id'].duplicated().sum())
```

```python
#Average order value per customer 

customer_aov = master_table_df.groupby('customer_unique_id').agg(
    total_spend=('total_value', 'sum'),
    total_orders=('order_id', 'nunique')
).reset_index()

customer_aov['avg_order_value'] = (
    customer_aov['total_spend'] / customer_aov['total_orders']
)

display(customer_aov.head())
```

```python
#Step 5: Exploratory Data Analysis (EDA)
#Perform structured analysis across the following dimensions:

#Customer Analysis
#New vs repeat customers 
#new_customer=(master_table_df['customer_purchase_freq']==1).sum()
#repeat_customer=(master_table_df['customer_purchase_freq']>1).sum()
master_table_df["customer_type"] = master_table_df["customer_purchase_freq"].apply(
    lambda x: "New" if x==1 else "Repeat")

customer_type_counts = master_table_df["customer_type"].value_counts()
print(customer_type_counts)
# Customer Type Distribution Graph

plt.figure(figsize=(8,5))

customer_type_counts.plot(
    kind='bar'
)

plt.title("New vs Repeat Customers")
plt.xlabel("Customer Type")
plt.ylabel("Count")

plt.xticks(rotation=0)

plt.show()
```

```python
#High-value vs low-value customers 

customer_value = (master_table_df.groupby('customer_unique_id')['customer_lifetime_value'].sum().reset_index())
#high_value_customer = customer_value[customer_value > customer_value.quantile(0.75)] 
#low_value_customer = customer_value[customer_value <= customer_value.quantile(0.75)]
display(master_table_df.head())
# Threshold (Top 25%)
threshold = customer_value['customer_lifetime_value'].quantile(0.75)

#Customer Segmentation
customer_value['value_segment'] = customer_value['customer_lifetime_value'].apply(
    lambda x: 'High Value' if x > threshold else 'Low Value'
)

print(customer_value.head())
plt.figure(figsize=(8,5))

sns.countplot(
    x='value_segment',
    data=customer_value
)

plt.title("High-value vs Low-value Customers")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")

plt.show()
```

```python
#Geographic distribution of customers ,SP seems to have a high customers base, highly ordered
customer_geography = (master_table_df.groupby('customer_state')['customer_unique_id'].nunique().sort_values(ascending=False))

print(customer_geography.head())
plt.figure(figsize=(10,6))

customer_geography.head(10).sort_values().plot(kind='barh')

plt.title("Top States by Customer Count")
plt.xlabel("Unique Customers")
plt.ylabel("Customer State")

plt.show()
```

```python
#Revenue and Order Analysis ,Monthly revenue trends ,Order volume trends ,Peak sales periods 

master_table_df['order_month'] = (master_table_df['order_purchase_timestamp'].dt.to_period('M'))
monthly_revenue = (master_table_df.groupby('order_month')['total_value'].sum())
monthly_orders = ( master_table_df.groupby('order_month')['order_id'].nunique())
peak_months = monthly_revenue.sort_values(ascending=False)

print(peak_months.head())
master_table_df['order_day'] = (master_table_df['order_purchase_timestamp'].dt.day_name())
orders_by_day = (master_table_df.groupby('order_day')['order_id'].nunique())

master_table_df['order_hour'] = (master_table_df['order_purchase_timestamp'].dt.hour)
orders_by_hour = (master_table_df.groupby('order_hour')['order_id'].nunique())

#monthly sale trend
plt.figure(figsize=(14,6))

monthly_orders.plot(
    marker='o',
    linewidth=2
)

plt.title("Monthly Order Volume Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()


#peak sale
plt.figure(figsize=(12,6))

peak_months.head(10).plot(kind='bar')

plt.title("Top Revenue Generating Months")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()

#day orederd trend
plt.figure(figsize=(10,5))

orders_by_day.plot(kind='bar')

plt.title("Orders by Day of Week")
plt.xlabel("Day")
plt.ylabel("Orders")

plt.xticks(rotation=45)

plt.show()
#peak sales hours
plt.figure(figsize=(12,5))

orders_by_hour.plot(
    marker='o'
)

plt.title("Orders by Hour")
plt.xlabel("Hour")
plt.ylabel("Orders")

plt.grid(True)

plt.show()


category_revenue = (master_table_df.groupby('product_category_name_english')['total_value'].sum().sort_values(ascending=False))
print(category_revenue.head(10))
plt.figure(figsize=(14,6))

category_revenue.head(10).plot(kind='bar')

plt.title("Revenue Contribution by Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python

#Product Analysis
#Top-selling product categories 
#Revenue contribution by category 
#Product demand distribution 


top_categories = (master_table_df.groupby('product_category_name_english')['order_id'].nunique().sort_values(ascending=False))
print(top_categories.head(10))
plt.figure(figsize=(14,6))

top_categories.head(10).plot(kind='bar')

plt.title("Top Selling Product Categories")
plt.xlabel("Product Category")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)

plt.show()

#bed_bath_table  product categories dominate sales volume,
#indicating strong customer demand and market preference.
#So we can get an insight about which products being ordered most

product_demand = (master_table_df.groupby('product_id')['order_id'].nunique())
plt.figure(figsize=(12,6))

sns.histplot(product_demand,bins=50)

plt.title("Product Demand Distribution")
plt.xlabel("Number of Orders")
plt.ylabel("Frequency")

plt.show()
print(product_demand.describe())
#A category may have high sales volume but low revenue due to lower product pricing, while another category may generate high revenue despite fewer orders because of higher average product value.
```

```python
#Seller Analysis
#Top-performing sellers 
#Seller contribution to revenue 
#Seller distribution 
#Tells which seller is selling the maximum number of items
top_sellers = (master_table_df.groupby('seller_id')['order_id'].nunique().sort_values(ascending=False))
print(top_sellers.head(10))
plt.figure(figsize=(14,6))

top_sellers.head(10).plot(kind='bar')

plt.title("Top Performing Sellers")
plt.xlabel("Seller ID")
plt.ylabel("Number of Orders")

plt.xticks(rotation=90)

plt.show()


seller_revenue = (master_table_df.groupby('seller_id')['total_value'].sum().sort_values(ascending=False)
)
plt.figure(figsize=(14,6))

seller_revenue.head(10).plot(kind='bar')

plt.title("Top Revenue Generating Sellers")
plt.xlabel("Seller ID")
plt.ylabel("Revenue")

plt.xticks(rotation=90)

plt.show()


seller_order_distribution = (master_table_df.groupby('seller_id')['order_id'].nunique())
plt.figure(figsize=(12,6))

sns.histplot(seller_order_distribution,bins=50)

plt.title("Seller Order Distribution")
plt.xlabel("Orders per Seller")
plt.ylabel("Frequency")

plt.show()
```

```python
#Identfy Average customer satisfaction.
#Whether most customers are happy or unhappy.

print(master_table_df['review_score'].describe())
review_distribution = (master_table_df['review_score'].value_counts().sort_index())

print(review_distribution)
plt.figure(figsize=(8,5))

sns.countplot(
    x='review_score',
    data=master_table_df
)

plt.title("Distribution of Review Scores")
plt.xlabel("Review Score")
plt.ylabel("Count")

plt.show()

master_table_df['delivery_days'] = (master_table_df['order_delivered_customer_date'] - master_table_df['order_purchase_timestamp']).dt.days
delivery_review = (master_table_df.groupby('review_score')['delivery_days'].mean())

print(delivery_review)

plt.figure(figsize=(10,6))

sns.boxplot(x='review_score',y='delivery_days',data=master_table_df)

plt.title("Delivery Time vs Review Score")
plt.xlabel("Review Score")
plt.ylabel("Delivery Days")

plt.show()


dissatisfied_customers = (master_table_df[master_table_df['review_score'] <= 2])
print(dissatisfied_customers['delivery_days'].describe())
low_rating_sellers = (dissatisfied_customers['seller_id'].value_counts().head(10))

print(low_rating_sellers)
plt.figure(figsize=(12,6))

low_rating_sellers.plot(kind='bar')

plt.title("Sellers with Highest Low Ratings")
plt.xlabel("Seller ID")
plt.ylabel("Low Rating Count")

plt.xticks(rotation=90)

plt.show()
```
