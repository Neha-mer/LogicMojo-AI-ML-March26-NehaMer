```python
Q13. Custom Feature Transformation using apply()

Context: Real ML requires custom logic transformation.

Dataset: Titanic

Tasks:

Create age_group: Child / Adult / Senior
Use .apply()
Compute survival rate per group
Interpret: Which segment has highest survival likelihood
```

```python
import seaborn as sns

data_raw=sns.load_dataset('titanic')

data=data_raw.copy()
data.head()

def age_group(age):
    if age<18:
        return "Child"
    elif age<60:
        return "Adult"
    else:
        return "Old"

data['age_group']=data['age'].apply(age_group)
data.head()
#Compute survival  per group
data.groupby('age_group')['survived'].mean()
#Childern has the highest survival rate
 
```

```python
Q14. Error Analysis Simulation (Model Evaluation Thinking)

Context: Understanding prediction errors is key in ML.

Tasks:

Generate 1000 random error values using NumPy
Compute: Mean Standard deviation
Identify outliers: Using mean ± 2*std
Explain: What high variance means in model performance
```

```python
import numpy as np

error_values=np.random.rand(1000)
#print(error_values)
mean_val=np.mean(error_values,axis=0)
std_dev=np.std(error_values,axis=0)
print(error_values)

lower_mean_val=mean_val-2*std_dev
upper_mean_val=mean_val+2*std_dev

print(lower_mean_val)
print(upper_mean_val)
#high varinace menas there is data which is inconsitenct and this data will not help much in prediction or formulating accurate decision
```
