```python
#3. Fraud Detection Rule Engine
#You are given:
#Task:

#Write a function is_fraud(transaction)
#If amount is greater than 100000 or location is "Unknown" mark as fraud
#Otherwise mark as safe
#Use lambda with filter or map
#Return only fraud transactions

transactions = [
    {"id": 101, "amount": 5000, "location": "India"},
    {"id": 102, "amount": 200000, "location": "Unknown"},
    {"id": 103, "amount": 1000, "location": "India"}
]


def is_fraud(transaction):

    if transaction["amount"] > 100000 or transaction["location"] == "Unknown":
        return True;

fraud_transactions = list(filter(is_fraud, transactions))

print(fraud_transactions)
```

```python
#6. Sequence Pattern Detection
#You are given:
#Task:

#Write a function detect_increasing_sequence(data)
#Find the longest continuous increasing sequence
#Return that sequence as a list
#Example output:
#[10, 20, 30]

data = [10, 20, 30, 25, 40, 50]
current_seq = [data[0]]
longest_seq = [data[0]]

def detect_increasing_sequence(i):
    global current_seq, longest_seq

    if data[i] > data[i - 1]:
        current_seq.append(data[i])
    else:
        current_seq = [data[i]]

    if len(current_seq) > len(longest_seq):
        longest_seq = current_seq[:]  

for i in range(1, len(data)):
    detect_increasing_sequence(i)

print(longest_seq)

#Shouldnt the queston ask first longest subseq cause ther 25,40, 50 as well

```

```python
#7. Dictionary-Based Recommendation System
#You are given:
#Task:

#Write a function recommend(user)
#Recommend items based on common items between users
#Do not recommend items already purchased by the user
#Use sets and dictionary iteration#
user_purchases = {
    "user1": ["apple", "banana"],
    "user2": ["banana", "orange"],
    "user3": ["apple", "orange"]
}

def recommend(user):
    my_items = set(user_purchases[user])
    other_items = set()

    for other_user, items in user_purchases.items():
        #user 2, user3
        if other_user != user:
            other_items.update(items) 
    
    #print(my_items)  
   # print(other_items)
    return other_items - my_items  

print(recommend("user1"))
print(recommend("user2"))
print(recommend("user3"))

    
    
```
