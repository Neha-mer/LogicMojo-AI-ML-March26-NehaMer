```python
#Calculates the average age of valid customers.
ages = [25, 42, -1, 34, None, 52, 17, -5, 29, None, 46]
valid_age=[]
avg_age=0 #avg=(sumof all the elements)/totalNoOfElements
sum=0
for i in ages:
    if i is not None and i >= 0:
        valid_age.append(i)
print(valid_age)
for i in valid_age:
    sum=sum+i
    length=len(valid_age)
    avg_age=sum/length
print(length)
print(sum)
print(avg_age)
```

```python
#Separates login hours into even-hour logins and odd-hour logins.
login_hours = [1, 3, 5, 7, 2, 6, 8, 4, 9, 10]
even_hours=[]
odd_hours=[]
for i in login_hours:
    if(i%2==0):
       even_hours.append(i)
    else:
        odd_hours.append(i)
print(f"even hours {even_hours}")
print(f"odd hours {odd_hours}")
```

```python
#Identifies login hours that fall into peak activity time (hours greater than 5).
login_hours = [1, 3, 5, 7, 2, 6, 8, 4, 9, 10]
peak_hours=[]
for i in login_hours:
    if i>5:
        peak_hours.append(i)
print(peak_hours)
```
